### 2.2. Arithmetic Operators (Tidyverse Perspective)

Arithmetic operators are fundamental base R operations. While `tidyverse` doesn't change how these operators work, they are commonly applied to numeric columns within `tibbles` (data frames) using functions like `dplyr::mutate()` to create new calculated columns.

```r
library(dplyr)
library(tibble)

# Define sample values in a tibble
math_data <- tibble(
  val_a = 15,
  val_b = 4
)

print("Original Values in Tibble:")
print(math_data)

# Perform arithmetic operations using mutate
results_data <- math_data %>%
  mutate(
    addition = val_a + val_b,
    subtraction = val_a - val_b,
    multiplication = val_a * val_b,
    division = val_a / val_b,
    modulo = val_a %% val_b,
    integer_division = val_a %/% val_b
  )

print("\nArithmetic Operations using dplyr::mutate():")
print(results_data)

# Example of power operation (can also be done in mutate)
power_example <- tibble(base = 2, exponent = 3) %>%
  mutate(result = base^exponent)

print("\nPower Operation Example:")
print(power_example)
```

### 6.1. Data Structures (Tidyverse Perspective)

The `tidyverse` primarily works with `tibbles` (enhanced data frames). While vectors, matrices, and lists are base R, they can often be converted to or from tibbles, or their contents can be manipulated to fit into a tibble structure. The `purrr` package (part of `tidyverse`) is excellent for working with lists.

```r
library(tibble)
library(dplyr)
library(purrr) # For list manipulation

# --- 1. Vector (converted to a tibble) ---
cat("\n--- Vector (as a tibble) ---\n")
vec_data_tbl <- tibble(value = c(1, 2, 3))
print("Vector as a tibble:")
print(vec_data_tbl)
print(paste("Type:", class(vec_data_tbl)[1]))
print(paste("Rows:", nrow(vec_data_tbl)))

# --- 2. Matrix (converted to a tibble) ---
cat("\n--- Matrix (converted to tibble) ---\n")
mat_orig <- matrix(1:8, ncol = 4, nrow = 2)
print("Original Matrix:")
print(mat_orig)
mat_data_tbl <- as_tibble(mat_orig, .name_repair = "universal") # Convert matrix to tibble
print("Matrix converted to tibble:")
print(mat_data_tbl)
print(paste("Type:", class(mat_data_tbl)[1]))
print(paste("Dimensions:", nrow(mat_data_tbl), "rows,", ncol(mat_data_tbl), "columns"))

# --- 3. List (manipulated with purrr) ---
cat("\n--- List (using purrr for inspection/manipulation) ---\n")
my_list_orig <- list(
  name = "Tirupati",
  age = 25,
  coords = c(13.62, 79.41),
  active = TRUE
)

print("Original List:")
print(my_list_orig)

# Use purrr::map_chr to get types of elements
list_element_types <- my_list_orig %>%
  map_chr(class)
print("List element types (using map_chr):")
print(list_element_types)

# Or convert list to tibble (if suitable structure using enframe)
list_to_tibble_enframe <- enframe(my_list_orig) # Converts named list to a two-column tibble
print("List converted to tibble (enframe):")
print(list_to_tibble_enframe)

# --- 4. Data Frame (already a tibble in tidyverse context) ---
cat("\n--- Data Frame (as a tibble) ---\n")
df_data_tbl <- tibble(id = 1:2, score = c(85, 92), city = c("Tambaram", "Chennai"))
print("Data Frame as a tibble:")
print(df_data_tbl)
print(paste("Type:", class(df_data_tbl)[1]))
print(paste("Dimensions:", nrow(df_data_tbl), "rows,", ncol(df_data_tbl), "columns"))
print(paste("Column Names:", paste(names(df_data_tbl), collapse = ", ")))
```

### 4.1. Logical Operators (Tidyverse Perspective)

Logical operators are often used within `dplyr::filter()` to combine multiple conditions, and within `dplyr::mutate()` to create new logical columns based on existing ones in `tibbles`.

```r
library(dplyr)
library(tibble)

# --- Scalar and Vector Logical Operations (remain base R) ---
cat("\n--- Scalar and Vector Logical Operations (still base R) ---\n")
p_log <- TRUE
q_log <- FALSE
cat("p & q (AND):     ", p_log & q_log, "\n")
cat("p | q (OR):      ", p_log | q_log, "\n")

x_vec_log <- c(TRUE, FALSE, TRUE, FALSE)
y_vec_log <- c(FALSE, TRUE, TRUE, FALSE)
cat("Vector x & y: ", x_vec_log & y_vec_log, "\n")


# --- Student Eligibility Filter (Tidyverse way with tibble, mutate, and filter) ---
student_eligibility_tbl <- tibble(
  student_id = 1:5,
  score = c(85, 72, 95, 68, 88),
  attendance_percent = c(95, 82, 98, 60, 90)
)

min_score_elig <- 75
min_attend_elig <- 80

cat("\n--- Student Eligibility (Tidyverse) ---\n")
print("Original Student Eligibility Tibble:")
print(student_eligibility_tbl)

eligible_students_tbl <- student_eligibility_tbl %>%
  mutate(
    meets_score = score >= min_score_elig,
    meets_attendance = attendance_percent >= min_attend_elig,
    is_eligible = meets_score & meets_attendance
  ) %>%
  filter(is_eligible) # Filter for students where is_eligible is TRUE

print("\nEligible Students (Tidyverse using mutate and filter):")
print(eligible_students_tbl)


# --- Image Processing: Multi-Condition Mask (Tidyverse way with tibble and mutate) ---
image_mask_tbl <- tibble(
  pixel_id = 1:5,
  is_bright = c(TRUE, FALSE, TRUE, FALSE, TRUE),
  is_edge = c(FALSE, TRUE, TRUE, FALSE, TRUE)
)

cat("\n--- Image Masking (Tidyverse) ---\n")
print("Original Image Mask Tibble:")
print(image_mask_tbl)

valid_pixels_tbl <- image_mask_tbl %>%
  mutate(is_valid_pixel = is_bright & is_edge) %>%
  filter(is_valid_pixel)

print("\nValid Pixels (Tidyverse using mutate and filter):")
print(valid_pixels_tbl)
```

### 3.2. Logical Operators (Tidyverse Perspective)

Logical operators are often used within `dplyr::filter()` to combine multiple conditions, and within `dplyr::mutate()` to create new logical columns based on existing ones.


```r
library(dplyr)

# Student Eligibility Filter using `filter()` and `mutate()`
student_eligibility_data <- tibble(
  student_id = 1:5,
  scores = c(85, 72, 95, 68, 88),
  attendance = c(95, 82, 98, 60, 90)
)

min_score <- 75
min_attend <- 80

eligible_students <- student_eligibility_data %>%
  mutate(
    is_passing_score = scores >= min_score,
    has_good_attendance = attendance >= min_attend,
    is_eligible = is_passing_score & has_good_attendance
  ) %>%
  filter(is_eligible)

print("\nStudent Eligibility (Tidyverse):")
print(eligible_students)

# Image Masking (Vision Example) using `mutate()`
image_mask_data <- tibble(
  pixel_id = 1:5,
  is_bright = c(TRUE, FALSE, TRUE, FALSE, TRUE),
  is_edge = c(FALSE, TRUE, TRUE, FALSE, TRUE)
)

valid_pixels_data <- image_mask_data %>%
  mutate(is_valid = is_bright & is_edge) %>%
  filter(is_valid)

print("\nImage Masking (Valid Pixels - Tidyverse):")
print(valid_pixels_data)
```
